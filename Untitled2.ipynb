* DATA CLEANING


Step 1: Load the Dataset

Load the dataset using pandas.

The dataset is stored at URL: https://gitlab.crio.do/me_notebook/me_jupyter_uberanalysis/-/raw/master/Uber_ride_analysis_dataset.csv .

Display the first few rows to get an overview of the data.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data_url = 'https://gitlab.crio.do/me_notebook/me_jupyter_uberanalysis/-/raw/master/Uber_ride_analysis_dataset.csv'

df = pd.read_csv(data_url)
print(df.head())
```

Step 2: Standardize Column Names

Print the column names of the dataset.

Standardize the column names:

Convert to lowercase

Replace spaces with underscores

```python
print(df.columns)
df.columns = df.columns.str.lower().str.replace(' ', '_')
print(df.columns)
```

Step 3: Handle Missing Values

Check for missing values in the dataset using .isnull().sum().

Handle them as follows:

Trip Status / Trip Cost: Drop rows where these are missing.

Driver ID: Fill with -1 to indicate unassigned drivers.

Payment Method: Fill with the most frequent value (mode).

Start Timestamp / Drop Timestamp:

Keep only if the trip was completed.

For cancelled or failed trips, missing timestamps are acceptable.

```python
print(df.isnull().sum())

df = df.dropna(subset=['trip_status', 'trip_cost'])
df['driver_id'] = df['driver_id'].fillna(-1)
df['payment_method'] = df['payment_method'].fillna(df['payment_method'].mode()[0])
mask = (df['trip_status'] == 'Completed') & ((df['start_timestamp'].isnull()) | (df['drop_timestamp'].isnull()))
df = df[~mask].reset_index()

print(df.isnull().sum())
```

Step 4: Convert Columns to Proper Types

Convert all timestamp columns to datetime format: request_timestamp, start_timestamp, Drop_timestamp

```python
concerned_columns = ['request_timestamp' , 'start_timestamp' , 'drop_timestamp']

for col in concerned_columns:
    df[col] = pd.to_datetime(df[col], errors = 'coerce')

df[concerned_columns].dtypes
```

Step 5: Identify Numeric Columns

Identify and list all numeric columns in the dataset.

```python
numeric_columns = df.select_dtypes(include=['number']).columns.tolist()
print(numeric_columns)

value_count = {col: df[col].nunique() for col in numeric_columns}
print(value_count)
```

Step 6: Outlier Detection and Treatment

Perform outlier detection on the following numeric fields: trip_cost and extra_tip

Steps:

Plot boxplots to visualize distribution with title as {column_name} distribution

Detect outliers using IQR method.

Cap outliers at the upper bound using np.where.

Re-plot the boxplot to verify changes with title as {column_name} distribution - Outliers Handled

```python
df['extra_tip'].plot(kind='box', vert=False)
plt.title('Extra Tip Distribution')

```

```python
df['trip_cost'].plot(kind='box', vert=False)
plt.title('Trip Cost Distribution')
```

```python
Q1 = df['trip_cost'].quantile(0.25)
Q3 = df['trip_cost'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

outliers = df[(df['trip_cost'] < lower_bound) | (df['trip_cost'] > upper_bound)]
print(outliers[['trip_cost']])
print('Number of Outliers:', ((df['trip_cost'] < lower_bound) | (df['trip_cost'] > upper_bound)).sum())
```

```python
Q1 = df['extra_tip'].quantile(0.25)
Q3 = df['extra_tip'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

outliers = df[(df['extra_tip'] < lower_bound) | (df['extra_tip'] > upper_bound)]
print(outliers[['extra_tip']])
print('Number of Outliers:', ((df['extra_tip'] < lower_bound) | (df['extra_tip'] > upper_bound)).sum())
```

```python
df['extra_tip'] = np.where(df['extra_tip']> upper_bound, upper_bound, df['extra_tip'])
df['extra_tip'].plot(kind='box', vert=False)
plt.title('Extra Tip Distribution- Outliers Handled')
plt.show()
```

```python
df['trip_cost'] = np.where(df['trip_cost']> upper_bound, upper_bound, df['trip_cost'])
df['trip_cost'].plot(kind='box', vert=False)
plt.title('Trip Cost Distribution- Outliers Handled')
plt.show()
```

Step 3: Extract Date and Time Components

From the relevant datetime column(s), extract:

Date (YYYY-MM-DD) (request_date, start_date, drop_date)

Day of the week (e.g., Monday, Tuesday) (request_day, start_day, drop_day)

Exact time (HH:MM:SS) (request_time, start_time, drop_time)

Hour of the day (0–23) (request_hour, start_hour, drop_hour)

Hint : Can use below functions for this

To extract date from the timestamp -

df[‘Col’].dt.date
To extract day name from the timestamp -

df[‘Col’].dt.day_name()
To extract time from the timestamp -

df[‘Col’].dt.time
To extract hour from the timestamp -

df[‘Col’].dt.hour

```python


df['request_date'] = df['request_timestamp'].dt.date
df['start_date'] = df['start_timestamp'].dt.date
df['drop_date'] = df['drop_timestamp'].dt.date

df['request_day'] = df['request_timestamp'].dt.day_name()
df['start_day'] = df['start_timestamp'].dt.day_name()
df['drop_day'] = df['drop_timestamp'].dt.day_name()

df['request_time'] = df['request_timestamp'].dt.time
df['start_time'] = df['start_timestamp'].dt.time
df['drop_time'] = df['drop_timestamp'].dt.time

df['request_hour'] = df['request_timestamp'].dt.hour
df['start_hour'] = df['start_timestamp'].dt.hour
df['drop_hour'] = df['drop_timestamp'].dt.hour
```

Step 4: Calculating the Ride delay.

Determine the delay between the ride request and actual trip start.

Add a column ride_delay reflecting this delay in hour value.

```python
df['ride_delay'] = (df['start_timestamp'] - df['request_timestamp']).dt.total_seconds()/3600
print(df.head())
```

Step 5: Determine Cancellation reasons

Some rides may have been canceled. Add a new column cancellation_reason with the following categories:

No Cabs – if no driver was assigned and the status reflects availability issues.

Passenger – if the user canceled the trip before pickup.

Driver – if a driver canceled the ride after accepting it.

Assign a cancellation reason to each trip based on the driver_id and trip_status columns using nested np.where() statements.

If the driver_id is -1 and the** trip status is 'No Cars Available'**, it indicates that no cab was assigned, so the cancellation reason is set to 'No Cabs'.

If the driver_id is -1 and the trip status is 'Trip Cancelled', it means the passenger canceled the trip before a driver was assigned, so the reason is 'Passenger'.

If a driver was assigned (driver_id not equal to -1) and the trip status is 'Trip Cancelled', it indicates that the driver canceled the trip, so the reason is 'Driver'.

For all other cases—where the trip was completed successfully—the cancellation reason is set to 'Trip Completed'.



```python
print(df.columns)
df['cancellation_reason'] = np.where((df['driver_id']==-1) & (df['trip_status'] == 'No Cars Available'), 'No Cabs',
                                    np.where((df['driver_id']==-1) & (df['trip_status'] == 'Trip Cancelled'), 'Passenger',
                                             np.where((df['driver_id']!=1) & (df['trip_status'] == 'Trip Cancelled'),'Driver', 'Trip Completed')))
print(df.head())
```
